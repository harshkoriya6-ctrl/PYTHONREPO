```python

import numpy as np


class DataAnalytics:

    def __init__(self):
        """Initialize the NumPy Analyzer and array storage."""
        self.arrays = {}


    def create_array(self):
        """Create 1D, 2D or 3D NumPy arrays."""
        print("\n1. 1D Array")
        print("2. 2D Array")
        print("3. 3D Array")

        choice = input("Enter choice: ")

        if choice == "1":
            data = list(map(int, input("Enter elements: ").split()))
            arr = np.array(data)

        elif choice == "2":
            rows = int(input("Enter rows: "))
            cols = int(input("Enter columns: "))

            data = list(map(int, input(
                f"Enter {rows * cols} elements: "
            ).split()))

            arr = np.array(data).reshape(rows, cols)

        elif choice == "3":
            layers = int(input("Enter layers: "))
            rows = int(input("Enter rows: "))
            cols = int(input("Enter columns: "))

            data = list(map(int, input(
                f"Enter {layers * rows * cols} elements: "
            ).split()))

            arr = np.array(data).reshape(layers, rows, cols)

        else:
            print("Invalid choice")
            return

        name = input("Enter array name: ")
        self.arrays[name] = arr

        print("Array created successfully:")
        print(arr)


    def select_array(self):
        """Select an existing array from storage."""
        if not self.arrays:
            print("No arrays found. Create an array first.")
            return None

        print("\nAvailable Arrays:")

        for name in self.arrays:
            print(name, ":", self.arrays[name])

        name = input("Enter array name: ")

        if name in self.arrays:
            return self.arrays[name]

        print("Array not found")
        return None
    

    def mathematical_operations(self):
        """Perform element-wise and matrix operations."""
        print("\n1. Addition")
        print("2. Subtraction")
        print("3. Multiplication")
        print("4. Division")
        print("5. Dot Product")
        print("6. Matrix Multiplication")

        choice = input("Enter choice: ")

        a = self.select_array()

        if a is None:
            return

        b = self.select_array()

        if b is None:
            return

        try:
            if choice == "1":
                print(np.add(a, b))

            elif choice == "2":
                print(np.subtract(a, b))

            elif choice == "3":
                print(np.multiply(a, b))

            elif choice == "4":
                if np.any(b == 0):
                    print("Division by zero is not allowed")
                else:
                    print(np.divide(a, b))

            elif choice == "5":
                print(np.dot(a, b))

            elif choice == "6":
                print(np.matmul(a, b))

            else:
                print("Invalid choice")

        except ValueError:
            print("Array shapes are incompatible")


    def combine_split(self):
        """Combine multiple arrays or split an array."""
        print("\n1. Combine Arrays")
        print("2. Split Array")

        choice = input("Enter choice: ")

        if choice == "1":
            a = self.select_array()

            if a is None:
                return

            b = self.select_array()

            if b is None:
                return

            try:
                result = np.concatenate((a, b), axis=0)
                print("Combined Array:")
                print(result)

            except ValueError:
                print("Arrays cannot be combined along this axis")

        elif choice == "2":
            arr = self.select_array()

            if arr is None:
                return

            if arr.shape[0] % 2 != 0:
                print("Array cannot be split into two equal parts")
                return

            result = np.split(arr, 2, axis=0)

            print("First Part:")
            print(result[0])

            print("Second Part:")
            print(result[1])

        else:
            print("Invalid choice")


    def search_sort_filter(self):
        """Search, sort and filter array elements."""
        print("\n1. Search")
        print("2. Sort Ascending")
        print("3. Sort Descending")
        print("4. Filter Greater Than a Value")

        choice = input("Enter choice: ")

        arr = self.select_array()

        if arr is None:
            return

        if choice == "1":
            value = int(input("Enter value to search: "))

            result = np.argwhere(arr == value)

            if result.size > 0:
                print("Value found at indices:")
                print(result)
            else:
                print("Value not found")

        elif choice == "2":
            print(np.sort(arr, axis=None))

        elif choice == "3":
            print(np.sort(arr, axis=None)[::-1])

        elif choice == "4":
            value = int(input("Enter filter value: "))

            result = arr[arr > value]

            print("Filtered Array:")
            print(result)

        else:
            print("Invalid choice")


    def aggregates_statistics(self):
        """Calculate aggregate and statistical values."""
        arr = self.select_array()

        if arr is None:
            return

        print("\nSum:", np.sum(arr))
        print("Mean:", np.mean(arr))
        print("Median:", np.median(arr))
        print("Standard Deviation:", np.std(arr))
        print("Variance:", np.var(arr))
        print("Minimum:", np.min(arr))
        print("Maximum:", np.max(arr))

    @classmethod
    def project_name(cls):
        """Display the project name."""
        print("\nNumPy Analyzer")


    @staticmethod
    def correlation(a, b):
        """Calculate the correlation coefficient of two arrays."""
        if a.size != b.size:
            print("Arrays must have the same number of elements")
            return

        result = np.corrcoef(a.flatten(), b.flatten())

        print("Correlation Coefficient:")
        print(result[0, 1])

```

```python

def main():
    
    obj = DataAnalytics()

    while True:

        print("""
===================================
       NUMPY ANALYZER
===================================
1. Create a NumPy Array
2. Mathematical Operations
3. Combine or Split Arrays
4. Search, Sort or Filter Arrays
5. Compute Aggregates and Statistics
6. Exit
""")

        choice = input("Enter your choice: ")

        if choice == "1":
            obj.create_array()

        elif choice == "2":
            obj.mathematical_operations()

        elif choice == "3":
            obj.combine_split()

        elif choice == "4":
            obj.search_sort_filter()

        elif choice == "5":
            obj.aggregates_statistics()

        elif choice == "6":
            print("Thank you for using NumPy Analyzer!")
            break

        else:
            print("Invalid choice. Try again.")


if __name__ == "__main__":
    main()
```
