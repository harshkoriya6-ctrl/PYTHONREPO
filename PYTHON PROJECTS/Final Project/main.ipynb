```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd
```

```python
class Titanic:
    def __init__(self):
        """
        This function loads the three csv files when class starts.
        Arguments taken: self which is the class object.
        Return value: none as it stores dataframes in class variables.
        """
        self.train = pd.read_csv("train.csv")
        self.test = pd.read_csv("test.csv")
        self.sub = pd.read_csv("gender_submission.csv")
        avg = self.train["Age"].mean()
        self.train["Age"] = self.train["Age"].fillna(avg)
        self.train["Embarked"] = self.train["Embarked"].fillna("S")
        print("files loaded")

    def check(self):
        """
        This function shows top rows shape and missing values of dataset.
        Arguments taken: self which holds the train dataframe.
        Return value: none because it only prints data on screen.
        """
        while True:
            print("\n")
            print("1 see top rows and shape")
            print("2 see missing values")
            print("3 see maths describe")
            print("4 back")
            ch = int(input("enter choice: "))
            if ch == 1:
                print(self.train.head())
                print(self.train.shape)
            elif ch == 2:
                print(self.train.isnull().sum())
            elif ch == 3:
                print(self.train.describe())
            elif ch == 4:
                break
            else:
                print("wrong input")

    def clean(self):
        """
        This function cleans missing age values and drops cabin column.
        Arguments taken: self which contains the main train dataframe.
        Return value: none as it updates the dataframe inside class.
        """
        while True:
            print("\n")
            print("1 fill age and embarked")
            print("2 drop cabin column")
            print("3 back")
            ans = int(input("enter option: "))
            if ans == 1:
                mid = self.train["Age"].mean()
                self.train["Age"] = self.train["Age"].fillna(mid)
                self.train["Embarked"] = self.train["Embarked"].fillna("S")
                print("filled empty values")
            elif ans == 2:
                if "Cabin" in self.train.columns:
                    self.train = self.train.drop(columns=["Cabin"])
                    print("cabin removed")
                else:
                    print("already removed")
            elif ans == 3:
                break
            else:
                print("wrong option")

    def eda(self):
        """
        This function calculates survival rates by gender class and fare.
        Arguments taken: self to use the cleaned train dataframe.
        Return value: none because it prints group calculations directly.
        """
        while True:
            print("\n")
            print("1 survival by gender and class")
            print("2 survival count total")
            print("3 average age and fare of survived")
            print("4 back")
            pick = int(input("enter number: "))
            if pick == 1:
                print(self.train.groupby("Sex")["Survived"].mean())
                print(self.train.groupby("Pclass")["Survived"].mean())
            elif pick == 2:
                print(self.train["Survived"].value_counts())
            elif pick == 3:
                print(self.train.groupby("Survived")["Age"].mean())
                print(self.train.groupby("Survived")["Fare"].mean())
            elif pick == 4:
                break
            else:
                print("try again")

    def matgraphs(self):
        """
        This function plots pie line scatter bar and hist charts using matplotlib.
        Arguments taken: self containing passenger columns for plotting.
        Return value: none as it opens graph windows on screen.
        """
        while True:
            print("\n")
            print("1 survival pie graph")
            print("2 class vs fare line plot")
            print("3 age vs fare scatter plot")
            print("4 gender bar graph")
            print("5 age histogram")
            print("6 back")
            num = int(input("enter graph number: "))
            if num == 1:
                vals = self.train["Survived"].value_counts()
                plt.figure()
                plt.pie(vals, labels=["Dead", "Survived"], autopct="%1.1f%%")
                plt.title("Survived vs Dead Pie Graph")
                plt.show()
            elif num == 2:
                line = self.train.groupby("Pclass")["Fare"].mean()
                plt.figure()
                plt.plot(line.index, line.values, marker="o", color="blue")
                plt.title("Average Fare by Class Line Plot")
                plt.show()
            elif num == 3:
                plt.figure()
                plt.scatter(self.train["Age"], self.train["Fare"], color="red")
                plt.title("Age vs Fare Scatter Plot")
                plt.show()
            elif num == 4:
                svals = self.train["Sex"].value_counts()
                plt.figure()
                plt.bar(svals.index, svals.values, color=["blue", "pink"])
                plt.title("Gender Bar Graph")
                plt.show()
            elif num == 5:
                plt.figure()
                plt.hist(self.train["Age"], bins=15, color="green")
                plt.title("Age Histogram")
                plt.show()
            elif num == 6:
                break
            else:
                print("wrong choice")

    def seagraphs(self):
        """
        This function creates countplot boxplot histplot and heatplot using seaborn.
        Arguments taken: self which holds the titanic dataset.
        Return value: none because it displays seaborn figures.
        """
        while True:
            print("\n")
            print("1 gender survival countplot")
            print("2 class survival countplot")
            print("3 age survival histplot")
            print("4 correlation heatplot")
            print("5 back")
            opt = int(input("enter seaborn option: "))
            if opt == 1:
                plt.figure()
                sns.countplot(x="Sex", hue="Survived", data=self.train)
                plt.title("Gender Survival Countplot")
                plt.show()
            elif opt == 2:
                plt.figure()
                sns.countplot(x="Pclass", hue="Survived", data=self.train)
                plt.title("Class Survival Countplot")
                plt.show()
            elif opt == 3:
                plt.figure()
                sns.histplot(data=self.train, x="Age", hue="Survived", bins=20)
                plt.title("Age Distribution Histplot")
                plt.show()
            elif opt == 4:
                plt.figure()
                nums = self.train[["Survived", "Pclass", "Age", "SibSp", "Parch", "Fare"]]
                sns.heatmap(nums.corr(), annot=True)
                plt.title("Correlation Heatplot")
                plt.show()
            elif opt == 5:
                break
            else:
                print("invalid option")

    def otherfiles(self):
        """
        This function merges and shows test and gender submission files.
        Arguments taken: self containing test and submission dataframes.
        Return value: none as it prints and plots merged data.
        """
        joined = pd.merge(self.test, self.sub, on="PassengerId")
        while True:
            print("\n")
            print("1 see merged test and submission data")
            print("2 plot test survival graph")
            print("3 back")
            c = int(input("enter choice: "))
            if c == 1:
                print(joined.head(10))
                print(joined["Survived"].value_counts())
            elif c == 2:
                plt.figure()
                sns.countplot(x="Pclass", hue="Survived", data=joined)
                plt.title("Test Data Survival")
                plt.show()
            elif c == 3:
                break
            else:
                print("wrong input")

    def docs(self):
        """
        This function prints documentation of all class functions at the end.
        Arguments taken: self to access all methods inside the class.
        Return value: none because it only prints docstrings on screen.
        """
        print(self.__init__.__doc__)
        print(self.check.__doc__)
        print(self.clean.__doc__)
        print(self.eda.__doc__)
        print(self.matgraphs.__doc__)
        print(self.seagraphs.__doc__)
        print(self.otherfiles.__doc__)
        print(self.docs.__doc__)


obj = Titanic()
while True:
    print("\n")
    print("1 check data")
    print("2 clean data")
    print("3 eda maths")
    print("4 matplotlib graphs")
    print("5 seaborn graphs")
    print("6 test and submission files")
    print("7 print docs and exit")
    mainch = int(input("enter main choice: "))
    if mainch == 1:
        obj.check()
    elif mainch == 2:
        obj.clean()
    elif mainch == 3:
        obj.eda()
    elif mainch == 4:
        obj.matgraphs()
    elif mainch == 5:
        obj.seagraphs()
    elif mainch == 6:
        obj.otherfiles()
    elif mainch == 7:
        obj.docs()
        print("program closed")
        break
    else:
        print("wrong choice")
```
